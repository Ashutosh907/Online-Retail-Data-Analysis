## 6. Data Visualizations

### 6.1 Monthly Revenue Trend

```python
import matplotlib.pyplot as plt
import pandas as pd

monthly_sales = (
    clean_df.groupby(['OrderYear', 'OrderMonth'])['Revenue']
    .sum()
    .reset_index()
)

monthly_sales['MonthYear'] = pd.to_datetime(
    monthly_sales['OrderYear'].astype(str) + '-' +
    monthly_sales['OrderMonth'].astype(str)
)

monthly_sales = monthly_sales.sort_values('MonthYear')

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales['MonthYear'],
    monthly_sales['Revenue'],
    marker='o'
)

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

### 6.2 Top 10 Countries by Revenue

```python
country_revenue = (
    clean_df.groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

country_revenue.plot(kind='bar')

plt.title('Top 10 Countries by Revenue')
plt.xlabel('Country')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### 6.3 Top 10 Products by Revenue

```python
product_revenue = (
    clean_df.groupby('Description')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

product_revenue.plot(kind='bar')

plt.title('Top 10 Products by Revenue')
plt.xlabel('Product')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### 6.5 Monthly Cancellation Rate

```python
monthly_orders = (
    df.groupby(['OrderYear', 'OrderMonth'])['InvoiceNo']
    .nunique()
)

monthly_cancelled = (
    df[df['InvoiceNo'].astype(str).str.startswith('C')]
    .groupby(['OrderYear', 'OrderMonth'])['InvoiceNo']
    .nunique()
)

monthly_cancellation_rate = (
    monthly_cancelled / monthly_orders * 100
).fillna(0)

monthly_cancellation_rate.index = pd.to_datetime(
    monthly_cancellation_rate.index.map(
        lambda x: f'{x[0]}-{x[1]:02d}-01'
    )
)

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_cancellation_rate.index,
    monthly_cancellation_rate.values,
    marker='o'
)

plt.title('Monthly Cancellation Rate')
plt.xlabel('Month')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
